```python
from pyspark.sql import SparkSession

# Create Spark Session
spark = SparkSession.builder.appName("Demo").getOrCreate()

# Sample data
data = [
    (1, "Ram", 25),
    (2, "Sita", 22),
    (3, "Krishna", 28)
]

# Create DataFrame
df = spark.createDataFrame(data, ["ID", "Name", "Age"])

# Display DataFrame
df.show()
```
